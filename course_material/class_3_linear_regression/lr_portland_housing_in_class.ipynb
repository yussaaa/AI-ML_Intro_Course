```python
df = pd.read_csv('../../data/temp_large/kaggle_housing_price/ex1data2.txt', header=None, names=['Size', 'Bedrooms', 'Price'])
df.head()
```

```python
def model(X, w, b) -> float:
    '''
    Simple linear regression model that predicts house prices based on X.
    X: int - The X feature of the house
    w: float - The weight (slope) of the model
    b: float - The bias (intercept) of the model

    return: float - The predicted price of the house

    '''
    return w * X + b

```

```python
df['prediction'] = df['Size'].apply(model, w=0.06, b=50)
```

```python
plt.scatter(df['Size'], df['Price'], label='Actual Price')
plt.scatter(df['Size'], df['prediction'], label='Predicted Price')
```

## Define the cost function and find the cost of each prediction & the entire model 

```python
def cost_func(pred, true):
    '''
    Calculate the cost (loss) between predicted and true values.
    pred: float - The predicted value
    true: float - The true value

    return: float - The cost (loss)
    '''
    return (pred - true) ** 2
```

```python
cost = df.apply(lambda row: cost_func(row['prediction'], row['Price']), axis=1)
```

```python
cost.sum() / (2 * len(cost))  # Average cost (loss) across all predictions
```
