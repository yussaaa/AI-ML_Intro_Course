```python
df.drop(columns=['sibsp', 'zero', 'zero.1',
       'zero.2', 'zero.3', 'zero.4', 'zero.5', 'zero.6', 'Parch', 'zero.7',
       'zero.8', 'zero.9', 'zero.10', 'zero.11', 'zero.12', 'zero.13',
       'zero.14', 'Pclass', 'zero.15', 'zero.16', 'Embarked', 'zero.17',
       'zero.18',], inplace=True)
```

```python
y = df['2urvived']
```

```python
y.value_counts()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
classifier = LogisticRegression(C=0.9)
```

```python
classifier.fit(X_train, y_train)
```

```python
classifier.score(X_test, y_test)
```

```python
predictions = classifier.predict(X_test)

predictions
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report

accuracy = accuracy_score(y_test, predictions)
precision = precision_score(y_test, predictions, average='weighted')
recall = recall_score(y_test, predictions, average='weighted')
f1 = f1_score(y_test, predictions, average='weighted')

accuracy, precision, recall, f1
```

```python
print(classification_report(y_test, predictions))
```

```python
classifier.coef_, classifier.intercept_
```

```python
from sklearn.metrics import log_loss

log_loss(y_test, predictions)
```

```python
def _log_loss(y_true, y_pred):
    epsilon = 1e-15
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)
    n = y_true.shape[0]
    loss = - (1/n) * np.sum(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))
    return loss
```

```python
def custom_log_loss(y_true, y_pred):
    return _log_loss(y_true, y_pred)
```

```python
custom_log_loss(y_test, predictions)
```
