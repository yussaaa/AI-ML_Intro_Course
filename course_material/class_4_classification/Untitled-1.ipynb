```python
import pandas as pd
import numpy as np
import matplotlib
matplotlib.use('TkAgg')
import matplotlib.pyplot as plt


cars = pd.read_csv(r"C:\Users\wypjj\Downloads\archive (3)\data.csv")



cars = cars.dropna(subset=["Engine HP", "MSRP"])

X = cars["Engine HP"].values
y = cars["MSRP"].values
n = len(y)

x_mean = np.mean(X)
y_mean = np.mean(y)


num = np.sum((X - x_mean) * (y - y_mean))
den = np.sum((X - x_mean) ** 2)
w = num / den
b = y_mean - w * x_mean


y_hat = w * X + b

J = (1 / (2 * n)) * np.sum((y_hat - y) ** 2)
print(f"Slope (w): {w:.2f}, Intercept (b): {b:.2f}")
print(f"Cost J(w, b): {J:,.2f}")

plt.figure(figsize=(7, 5))
plt.scatter(X, y, s=60, c='tab:blue', edgecolors='black', alpha=0.7, label="Actual Price")
plt.scatter(X, y_hat, s=60, c='crimson', edgecolors='black', alpha=0.7, label="Predicted Price")

for i in range(n):
    plt.arrow(X[i], y_hat[i], 0, y[i] - y_hat[i],
              length_includes_head=True,
              head_width=5, head_length=5000,
              color='gray', alpha=0.4)

plt.title("Car Prices: Actual vs Predicted")
plt.xlabel("Engine HP")
plt.ylabel("Price (USD)")
plt.grid(True, linestyle='--', linewidth=0.5, alpha=0.6)
plt.legend(["Actual Price", "Predicted Price"])
plt.tight_layout()

mng = plt.get_current_fig_manager()
try:
    mng.window.state('normal')
    mng.resize(800, 600)
except Exception:
    pass

plt.show()
```

```python
%matplotlib inline

import sys
print("Running from:", sys.executable)  # sanity check: venv path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Load and clean your dataset
cars = pd.read_csv(r"C:\Users\wypjj\Downloads\archive (3)\data.csv")
cars = cars.dropna(subset=["Engine HP", "MSRP"])

# Prepare features (X) and target (y)
X = cars[["Engine HP"]].to_numpy()
y = cars["MSRP"].to_numpy()

# Fit sklearn LinearRegression
reg = LinearRegression().fit(X, y)
y_hat = reg.predict(X)

# Metrics
print(f"R² score: {reg.score(X, y):.4f}")
print(f"coef_: {reg.coef_[0]:.2f}")
print(f"intercept_: {reg.intercept_:.2f}")
print(f"MSE: {mean_squared_error(y, y_hat):,.2f}")

# --- Plot like your gold graph ---
plt.figure(figsize=(7, 5))

# Actual prices in blue
plt.scatter(X, y, s=60, c='tab:blue', edgecolors='black', alpha=0.7, label="Actual Price")

# Predicted prices in red
plt.scatter(X, y_hat, s=60, c='crimson', edgecolors='black', alpha=0.7, label="Predicted Price")

# Residual “loss” lines
for xi, yi, yh in zip(X.ravel(), y, y_hat):
    plt.plot([xi, xi], [yh, yi], color='gray', linestyle='-', linewidth=1, alpha=0.4)

plt.title("Car Prices: Actual vs Predicted")
plt.xlabel("Engine HP")
plt.ylabel("Price (USD)")
plt.grid(True, linestyle='--', linewidth=0.5, alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
```
