```python
import numpy as np 
from matplotlib import pyplot as plt
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
def plot(f):
    z = np.linspace(-10, 10, 100)
    plt.plot(z, f(z))
    plt.title('Sigmoid Function')
    plt.xlabel('z')
    plt.ylabel('sigmoid(z)')
    plt.show()
```

```python
def line(x, w=1, b=4):
    return w * x + b   


plot(line)
```

```python
plot(sigmoid)
```
